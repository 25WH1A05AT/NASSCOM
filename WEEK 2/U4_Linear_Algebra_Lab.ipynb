```python
# 2B. NORMS AND COSINE SIMILARITY

import numpy as np
import numpy.linalg as la

a = np.array([1, 2, 3])
b = np.array([4, 0, 1])

print('L2 norm ||a||_2:', la.norm(a))
print('L1 norm ||a||_1:', la.norm(a, 1))
print('Linf norm:', la.norm(a, np.inf))

def cosine(u, v):
    return np.dot(u, v) / (la.norm(u) * la.norm(v))

print('cosine(a, b):', round(cosine(a, b), 3))
```

```python
# LAB EXERCISE 2: L1/L2 NORMS AND COSINE SIMILARITY

import numpy as np
import numpy.linalg as la

def cosine(u, v):
    return np.dot(u, v) / (la.norm(u) * la.norm(v))

pairs = [
    (np.array([1, 0, 1]), np.array([1, 0, 1])),
    (np.array([1, 2, 3]), np.array([3, 2, 1])),
    (np.array([2, 0, 0]), np.array([0, 5, 0]))
]

for i, (u, v) in enumerate(pairs, start=1):
    l1_norm = la.norm(u, 1)
    l2_norm = la.norm(u)
    similarity = cosine(u, v)

    print(f"Pair {i}:")
    print("L1 norm:", l1_norm)
    print("L2 norm:", l2_norm)
    print("Cosine similarity:", round(similarity, 3))
    print()

# Most similar pair = Pair 1
```

```python
# 3A. CORE MATRIX OPERATIONS

A = np.array([[2., 1.],
              [1., 3.]])
B = np.array([[1., 0.],
              [4., 2.]])

print('A @ B (matrix multiply):\n', A @ B)
print('A.T   (transpose):\n', A.T)
print('inverse(A):\n', la.inv(A))
print('trace(A) = sum of diagonal:', np.trace(A))
```

```python
# 🔹 3B. SPECIAL MATRICES: identity, symmetric, orthogonal

I = np.eye(2)
print('Identity I:\n', I)
print('A @ inv(A) == I ?', np.allclose(A @ la.inv(A), I))


print('A symmetric?', np.allclose(A, A.T))

theta = np.radians(30)
Q = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
print('Q orthogonal?', np.allclose(Q.T @ Q, I))
```

```python
# LAB EXERCISE 3: INVERSE, TRACE, AND MATRIX TESTS

import numpy as np

M = np.array([
    [4., 2.],
    [2., 3.]
])

P = np.array([
    [0., -1.],
    [1.,  0.]
])

inverse_M = np.linalg.inv(M)
trace_M = np.trace(M)

print("Inverse of M:\n", inverse_M)
print("Trace of M:", trace_M)

identity_check = np.allclose(M @ inverse_M, np.eye(2))
print("M @ inverse(M) is identity:", identity_check)

is_symmetric = np.allclose(M, M.T)
is_orthogonal = np.allclose(P.T @ P, np.eye(2))

print("M is symmetric:", is_symmetric)
print("P is orthogonal:", is_orthogonal)
```

```python
# LAB EXERCISE 4: APPLY TWO TRANSFORMATIONS AND PLOT

import numpy as np
import matplotlib.pyplot as plt

tri = np.array([
    [0, 1, 0.5],
    [0, 0, 1.0]
])

# 1. Scaling matrix

scale = np.array([
    [2, 0],
    [0, 0.5]
])


# 2. Rotation matrix

angle = np.radians(45)

rotation = np.array([
    [np.cos(angle), -np.sin(angle)],
    [np.sin(angle),  np.cos(angle)]
])


# 3. Apply transformations and plot

scaled_tri = scale @ tri
transformed_tri = rotation @ scaled_tri

def close_loop(points):
    return np.column_stack((points, points[:, 0]))

original_plot = close_loop(tri)
transformed_plot = close_loop(transformed_tri)

plt.figure(figsize=(7, 6))

plt.plot(
    original_plot[0],
    original_plot[1],
    'o-',
    label='Original Triangle'
)

plt.plot(
    transformed_plot[0],
    transformed_plot[1],
    's-',
    label='Transformed Triangle'
)

plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Original vs Transformed Triangle')
plt.legend()
plt.grid(True)
plt.axis('equal')
plt.show()
```

```python
#5A. COMPUTE EIGENVALUES & EIGENVECTORS


A = np.array([[2., 0.],
              [0., 3.]])

vals, vecs = la.eig(A)     # vals = eigenvalues, vecs columns = eigenvectors
print('Eigenvalues  (lambda):', vals)
print('Eigenvectors (columns):\n', vecs)
```

```python
# 5B. VERIFY A v = lambda v

for i in range(len(vals)):
    v = vecs[:, i]
    lhs = A @ v
    rhs = vals[i] * v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
```

```python
# LAB EXERCISE 5: EIGENVALUES, EIGENVECTORS AND VERIFICATION

import numpy as np
import numpy.linalg as la

C = np.array([
    [4., 1.],
    [2., 3.]
])

# 1. Eigenvalues and eigenvectors

vals, vecs = la.eig(C)

# 2. Print eigenvalues

print("Eigenvalues:", vals)

# 3. Verify C v = lambda v for the first eigenvector

v = vecs[:, 0]
lhs = C @ v
rhs = vals[0] * v

print("First eigenvector:", v)
print("C @ v:", lhs)
print("lambda * v:", rhs)
print("C @ v == lambda * v:", np.allclose(lhs, rhs))
```

```python
# LAB EXERCISE 6: SOLVE A SYSTEM AND COMPARE EMBEDDINGS

import numpy as np
import numpy.linalg as la

A2 = np.array([
    [3., 2.],
    [1., 4.]
])

b2 = np.array([7., 9.])

cat = np.array([0.9, 0.8, 0.1])
dog = np.array([0.85, 0.7, 0.2])
car = np.array([0.1, 0.2, 0.95])


# 1. Solve A2 x = b2

x = la.solve(A2, b2)
print("Solution x:", x)


# 2. Rank of A2

rank_A2 = la.matrix_rank(A2)
print("Rank of A2:", rank_A2)


# 3. Cosine similarity

def cosine(u, v):
    return np.dot(u, v) / (la.norm(u) * la.norm(v))

cat_dog = cosine(cat, dog)
cat_car = cosine(cat, car)

print("Cosine similarity cat vs dog:", round(cat_dog, 3))
print("Cosine similarity cat vs car:", round(cat_car, 3))

if cat_dog > cat_car:
    print("Cat is more similar to dog.")
else:
    print("Cat is more similar to car.")
```

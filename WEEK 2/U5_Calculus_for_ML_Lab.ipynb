```python
# 1A. NUMERICAL DERIVATIVE

def f(x):
    return x**2

h = 1e-6
slope_at_3 = (f(3 + h) - f(3)) / h

print("Numerical f'(3):", round(slope_at_3, 4))
```

```python
import sympy as sp

x = sp.symbols('x')

expr = x**2

deriv = sp.diff(expr, x)

print('d/dx (x**2) =', deriv)

print('Symbolic f\'(3) =', deriv.subs(x, 3))
```

```python
# LAB EXERCISE 1: NUMERICAL VS SYMBOLIC DERIVATIVE

import sympy as sp

x = sp.symbols('x')

def g(x):
    return x**3 + 2*x

h = 1e-6
numerical_derivative = (g(2 + h) - g(2)) / h

print("Numerical derivative at x = 2:", numerical_derivative)

expr = x**3 + 2*x
symbolic_derivative = sp.diff(expr, x)

print("Symbolic derivative:", symbolic_derivative)

symbolic_at_2 = symbolic_derivative.subs(x, 2)

print("Symbolic derivative at x = 2:", symbolic_at_2)
print("Values match:", abs(numerical_derivative - float(symbolic_at_2)) < 1e-5)
```

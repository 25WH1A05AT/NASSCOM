```python
import os
import datetime
import random

print("--- Intermediate Python Core Exercise ---")
print("This exercise covers Python fundamentals, data structures, functions,")
print("core libraries (os, datetime, random), file handling, and exception handling.")
print("Estimated completion time: 3-4 hours.")
print("------------------------------------------")

# --- Global Data for the Exercise (Simulated Data) ---
# We'll imagine this is sensor data from various devices.
DEVICE_TYPES = [
    'Temperature_Sensor',
    'Humidity_Sensor',
    'Pressure_Sensor',
    'Light_Sensor'
]

LOCATIONS = [
    'Lab1',
    'Lab2',
    'OfficeA',
    'OfficeB',
    'Warehouse'
]

READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0),   # Celsius
    'Humidity_Sensor': (30.0, 80.0),      # Percentage
    'Pressure_Sensor': (980.0, 1030.0),   # hPa
    'Light_Sensor': (100, 1000)           # Lux
}


# ============================================================
# TASK 1: Python Fundamentals, Lists, Tuples, Sets, Dictionaries
# ============================================================

print("\n--- Task 1: Data Generation and Basic Structures ---")


# ------------------------------------------------------------
# 1.1 Generate 100 simulated sensor readings
# ------------------------------------------------------------

raw_sensor_data = []

# Use a starting timestamp
start_time = datetime.datetime.now().replace(microsecond=0)

for i in range(100):
    timestamp = start_time + datetime.timedelta(minutes=i)

    timestamp_str = timestamp.strftime('%Y-%m-%d %H:%M:%S')

    device_id = f"DEV{i + 1:03d}"

    device_type = random.choice(DEVICE_TYPES)

    location = random.choice(LOCATIONS)

    min_value, max_value = READING_RANGES[device_type]

    value = round(random.uniform(min_value, max_value), 2)

    reading = (
        timestamp_str,
        device_id,
        device_type,
        location,
        value
    )

    raw_sensor_data.append(reading)

print(f"Generated {len(raw_sensor_data)} sensor readings.")
print("First 5 readings:")

for reading in raw_sensor_data[:5]:
    print(reading)


# ------------------------------------------------------------
# 1.2 Create temperature_readings
# ------------------------------------------------------------

temperature_readings = []

for timestamp, device_id, device_type, location, value in raw_sensor_data:

    if device_type == 'Temperature_Sensor':

        temperature_reading = {
            'timestamp': timestamp,
            'device_id': device_id,
            'location': location,
            'temperature_celsius': value
        }

        temperature_readings.append(temperature_reading)

print(
    f"\nNumber of temperature readings: "
    f"{len(temperature_readings)}"
)

print("First 3 temperature readings:")

for reading in temperature_readings[:3]:
    print(reading)


# ------------------------------------------------------------
# 1.3 Create unique_locations set
# ------------------------------------------------------------

unique_locations = set()

for reading in raw_sensor_data:
    location = reading[3]
    unique_locations.add(location)

print("\nUnique locations:")
print(unique_locations)


# ------------------------------------------------------------
# 1.4 Create device_stats dictionary
# ------------------------------------------------------------

device_stats = {}

# Initialize each device type
for device_type in DEVICE_TYPES:
    device_stats[device_type] = (float('inf'), float('-inf'))

# Find minimum and maximum values
for timestamp, device_id, device_type, location, value in raw_sensor_data:

    current_min, current_max = device_stats[device_type]

    new_min = min(current_min, value)
    new_max = max(current_max, value)

    device_stats[device_type] = (new_min, new_max)

print("\nDevice statistics:")

for device_type, stats in device_stats.items():
    print(
        f"{device_type}: "
        f"Min = {stats[0]}, "
        f"Max = {stats[1]}"
    )


# ============================================================
# TASK 1.5: Advanced Indexing and Slicing
# ============================================================

print("\n--- Task 1.5: Advanced Indexing and Slicing ---")

sample_string = "PythonProgrammingIsFun"


# ------------------------------------------------------------
# 1.5.1 List Slicing
# ------------------------------------------------------------

# 10th to 20th elements inclusive
# Python index 9 through 19
sliced_readings_1 = raw_sensor_data[9:20]

# Every 5th reading starting from beginning
sliced_readings_2 = raw_sensor_data[::5]

print("\nSliced readings 1:")
print(sliced_readings_1)

print("\nSliced readings 2:")
print(sliced_readings_2)


# ------------------------------------------------------------
# 1.5.2 String Slicing
# ------------------------------------------------------------

# "Programming" starts at index 6 and ends before index 17
substring_1 = sample_string[6:17]

# Reverse "Python"
# "Python" is sample_string[0:6]
substring_2 = sample_string[0:6][::-1]

print("\nSubstring 1:")
print(substring_1)

print("\nSubstring 2:")
print(substring_2)


# ------------------------------------------------------------
# 1.5.3 Mixed Slicing
# ------------------------------------------------------------

# 5th reading = index 4
fifth_reading_device_type = raw_sensor_data[4][2]

# Last 6 characters
device_type_suffix = fifth_reading_device_type[-6:]

print("\n5th reading device type:")
print(fifth_reading_device_type)

print("\nDevice type suffix:")
print(device_type_suffix)


# ============================================================
# TASK 2: Functions
# ============================================================

print("\n--- Task 2: Function Definitions and Usage ---")


# ------------------------------------------------------------
# 2.1 get_readings_by_location()
# ------------------------------------------------------------

def get_readings_by_location(readings_list, target_location):
    """
    Return all readings belonging to target_location.
    """

    matching_readings = []

    for reading in readings_list:

        location = reading[3]

        if location == target_location:
            matching_readings.append(reading)

    return matching_readings


# Test function
lab1_readings = get_readings_by_location(
    raw_sensor_data,
    'Lab1'
)

print(f"\nNumber of readings from Lab1: {len(lab1_readings)}")


# ------------------------------------------------------------
# 2.2 calculate_average_temperature()
# ------------------------------------------------------------

def calculate_average_temperature(temp_data_list):
    """
    Calculate the average temperature.

    Returns 0.0 if the list is empty.
    """

    if not temp_data_list:
        return 0.0

    total_temperature = 0.0

    for reading in temp_data_list:
        total_temperature += reading['temperature_celsius']

    average = total_temperature / len(temp_data_list)

    return average


# Test function
average_temperature = calculate_average_temperature(
    temperature_readings
)

print(
    f"Average temperature: "
    f"{average_temperature:.2f} °C"
)


# ------------------------------------------------------------
# 2.3 generate_report_line()
# ------------------------------------------------------------

def generate_report_line(reading_dict):
    """
    Convert a temperature reading dictionary
    into a formatted report line.
    """

    return (
        f"[{reading_dict['timestamp']}] "
        f"Device {reading_dict['device_id']} "
        f"at {reading_dict['location']}: "
        f"{reading_dict['temperature_celsius']}°C"
    )


# Test function if data exists
if temperature_readings:

    sample_report_line = generate_report_line(
        temperature_readings[0]
    )

    print("\nSample report line:")
    print(sample_report_line)


# ============================================================
# TASK 3: Core Python Libraries
# ============================================================

print("\n--- Task 3: Core Library Usage ---")


# ------------------------------------------------------------
# 3.1 Create sensor_reports directory
# ------------------------------------------------------------

reports_dir = 'sensor_reports'

if not os.path.exists(reports_dir):

    os.makedirs(reports_dir)

    print(
        f"\nDirectory '{reports_dir}' created."
    )

else:

    print(
        f"\nDirectory '{reports_dir}' already exists."
    )


# ------------------------------------------------------------
# 3.2 Calculate duration
# ------------------------------------------------------------

if len(raw_sensor_data) > 1:

    first_timestamp_str = raw_sensor_data[0][0]

    last_timestamp_str = raw_sensor_data[-1][0]

    date_format = '%Y-%m-%d %H:%M:%S'

    first_dt = datetime.datetime.strptime(
        first_timestamp_str,
        date_format
    )

    last_dt = datetime.datetime.strptime(
        last_timestamp_str,
        date_format
    )

    duration = last_dt - first_dt

    days = duration.days

    total_seconds = int(
        duration.total_seconds()
    )

    hours = (total_seconds // 3600) % 24

    minutes = (total_seconds % 3600) // 60

    print(
        f"\nDuration of data collection: "
        f"{days} days, "
        f"{hours} hours, "
        f"{minutes} minutes."
    )

else:

    print(
        "\nNot enough data to calculate duration."
    )


# ------------------------------------------------------------
# 3.3 Generate random values
# ------------------------------------------------------------

random_values = [
    random.randint(100, 200)
    for _ in range(5)
]

print("\nRandom values:")
print(random_values)


# ============================================================
# TASK 4: File Handling
# ============================================================

print("\n--- Task 4: File I/O ---")


# ------------------------------------------------------------
# 4.1 Write temperature readings to file
# ------------------------------------------------------------

report_filename = os.path.join(
    reports_dir,
    'temperature_report.txt'
)

try:

    with open(
        report_filename,
        'w',
        encoding='utf-8'
    ) as f:

        for reading_dict in temperature_readings:

            line = generate_report_line(
                reading_dict
            )

            f.write(line + '\n')

    print(
        f"\nTemperature report written to "
        f"'{report_filename}'."
    )

except IOError as e:

    print(
        f"\nError writing file "
        f"'{report_filename}': {e}"
    )


# ------------------------------------------------------------
# 4.2 Read report back into a list
# ------------------------------------------------------------

read_report_lines = []

try:

    with open(
        report_filename,
        'r',
        encoding='utf-8'
    ) as f:

        read_report_lines = [
            line.strip()
            for line in f
        ]

    print(
        f"\nSuccessfully read "
        f"{len(read_report_lines)} lines "
        f"from the report."
    )

    print("\nFirst 3 lines of the report:")

    for line in read_report_lines[:3]:
        print(line)

except FileNotFoundError:

    print(
        f"\nFile '{report_filename}' "
        f"was not found."
    )

except IOError as e:

    print(
        f"\nError reading file "
        f"'{report_filename}': {e}"
    )


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n--- Exercise Complete! ---")

print(
    "Great job completing the intermediate "
    "Python exercise."
)

print(
    "Review your code and outputs to ensure "
    "correctness and understanding of all concepts."
)

print(
    "Consider adding more complex scenarios "
    "or user inputs for further practice."
)


# ============================================================
# OPTIONAL SUMMARY OF RESULTS
# ============================================================

print("\n--- Final Results Summary ---")

print(
    f"Total sensor readings: "
    f"{len(raw_sensor_data)}"
)

print(
    f"Temperature readings: "
    f"{len(temperature_readings)}"
)

print(
    f"Unique locations: "
    f"{len(unique_locations)}"
)

print(
    f"Average temperature: "
    f"{average_temperature:.2f} °C"
)

print(
    f"Random values: "
    f"{random_values}"
)

print(
    f"Report file: "
    f"{report_filename}"
)
```

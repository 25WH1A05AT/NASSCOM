```python
# 🔹 1. LISTS

items = ['mango', 'orange', 'grape', 'mango']
print("Original list:", items)

print("First item:", items[0])
```

```python
items.append('banana')
print("After appending:", items)
```

```python
items.remove('grape')
print("After removing grape:", items)

# List length
print("Number of fruits:", len(items))
```

```python
# 2. TUPLES

dimensions = (10, 20, 30)
print("Tuple:", dimensions)

```

```python

# Accessing elements
print("Second element:", dimensions[1])
```

```python
# 3.SETS
unique_numbers = {1, 2, 3, 2, 1}
print("Set (no duplicates):", unique_numbers)
```

```python
# Adding elements
unique_numbers.add(4)
print("After adding 4:", unique_numbers)

```

```python
unique_numbers.remove(2)
print("After removing 2:", unique_numbers)

```

```python
set_a = {1, 2, 3}
set_b = {3, 4, 5}
print("Union:", set_a | set_b)
print("Intersection:", set_a & set_b)
print("Difference (A - B):", set_a - set_b)
```

```python
# 🔹 4. DICTIONARIES

student = {
    'name': 'shreeya',
    'age': 18,
    'course': 'Python'
}
print("Dictionary:", student)

```

```python
student['grade'] = 'A'
print("After adding grade:", student)


del student['age']
print("After deleting age:", student)


for key, value in student.items():
    print(f"{key}: {value}")
```

```python
def greet():
    print("Hello! Welcome to NASSCOM.")

greet()
```

```python
# 🔹 2. FUNCTION WITH PARAMETERS

def greet_user(name):
    print(f"Hello, {name}!")

greet_user("SHREEYA")
greet_user("MANASHWII")
```

```python

# 🔹 3. FUNCTION WITH DEFAULT PARAMETERS

def greet_user(name="Guest"):
    print(f"Welcome, {name}!")

greet_user()         # Uses default
greet_user("SHREEYA")
```

```python
# 🔹 5. USING *args (VARIABLE NUMBER OF POSITIONAL ARGUMENTS)

def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)

print_names("Alice", "bob", "Charlie")
```

```python

# 🔹 6. USING **kwargs (VARIABLE NUMBER OF KEYWORD ARGUMENTS)

def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")

print_info(name="SHREEYA", age=25, city="HYDERABAD")
```

```python
# 🔹 7. COMBINING *args AND **kwargs

def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_details(1, 2, 3, name="Ravi", city="Mumbai")
```

```python
# 🔹 8. FUNCTION NESTING (CALL ONE FUNCTION INSIDE ANOTHER)

def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 25:", square(25))
```
